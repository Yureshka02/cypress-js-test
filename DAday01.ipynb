```python
newdf = df.fillna(df.total_bedrooms.mean().round())
```

```python
newdf.isnull().sum()
```

```python
#Histogram
import matplotlib.pyplot as plt

plt.hist(newdf['median_house_value'], bins=20, color='lightblue')
plt.title('Distribution of Median House Value')
plt.xlabel('Median House Value')
plt.ylabel('Frequency')
plt.show()
```

```python
import matplotlib.pyplot as plt

newdf.plot(kind='scatter', x='longitude', y='latitude',
           alpha=0.5, c='median_house_value', cmap='jet', colorbar=True)
plt.title('Geo Housing Prices')
plt.xlabel('Longitude')
plt.ylabel('latitude')
plt.show()
```
